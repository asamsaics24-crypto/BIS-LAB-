```python
import numpy as np
import random


ETc = np.array([4.2,4.5,4.8,5,5.2,5.1,5,4.8,4.6,4.5,
                4.4,4.6,4.9,5.1,5.3,5.4,5.2,5,4.8,4.7,
                4.5,4.4,4.6,4.9,5.1,5.2,5,4.8,4.6,4.4])


rain = np.array([0,0,2,0,0,5,0,0,0,3,0,0,0,0,4,
                 0,0,0,2,0,0,0,3,0,0,0,0,0,0,0])


days, FC, WP, initial = 10, 10, 8, 25
pop_size, generations = 100, 200


def simulate(irrigation):
    sw, stress = initial, 0
    history = []


    for i in range(days):
        sw += rain[i] + irrigation[i] - ETc[i]
        sw = min(sw, FC)
        if sw < WP:
            stress += WP - sw
        sw = max(0, sw)
        history.append(sw)


    return np.array(history), stress


def fitness(x):
    _, stress = simulate(x)
    return np.sum(x) + 1000 * stress


pop = np.random.uniform(0, 15, (pop_size, days))
best = None


for g in range(generations):
    scores = np.array([fitness(x) for x in pop])
    best = pop[np.argmin(scores)].copy()


    parents = pop[np.argsort(scores)[:pop_size//2]]
    new = [best]


    while len(new) < pop_size:
        p1, p2 = random.choices(parents, k=2)
        point = random.randint(1, days-1)
        child = np.r_[p1[:point], p2[point:]]


        for i in range(days):
            if random.random() < 0.05:
                child[i] += np.random.normal(0, 2)


        new.append(np.clip(child, 0, 15))


    pop = np.array(new)


soil, stress = simulate(best)


print("Total irrigation:", round(np.sum(best), 2), "mm")
print("Water stress:", round(stress, 2))
print("Fitness:", round(fitness(best), 2))


for i in range(days):
    print(f"Day {i+1}: Irrigation={best[i]:.2f}, Soil={soil[i]:.2f}")
```
