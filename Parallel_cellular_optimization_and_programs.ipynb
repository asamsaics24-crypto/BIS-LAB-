```python

import numpy as np
import random

GRID_SIZE = 10
MAX_ITERATIONS = 5

def initialize_grid():
    grid = np.random.choice([0, 1], size=(GRID_SIZE, GRID_SIZE),
                            p=[0.7, 0.3])
    return grid

def fitness(grid):
    congestion = sum(np.sum(row) ** 2 for row in grid)
    return 1 / (1 + congestion)

def update_grid(grid):
    new_grid = grid.copy()
    for i in range(GRID_SIZE):
        for j in range(GRID_SIZE):
            if grid[i, j] == 1:
                neighbors = []
                for x, y in [(i-1,j), (i+1,j),
                             (i,j-1), (i,j+1)]:
                    if 0 <= x < GRID_SIZE and 0 <= y < GRID_SIZE:
                        if grid[x, y] == 0:
                            neighbors.append((x, y))
                if neighbors:
                    x, y = random.choice(neighbors)
                    new_grid[x, y] = 1
                    new_grid[i, j] = 0
    return new_grid

grid = initialize_grid()
best_fitness = 0
best_grid = grid.copy()

for iteration in range(MAX_ITERATIONS):
    score = fitness(grid)

    if score > best_fitness:
        best_fitness = score
        best_grid = grid.copy()

    grid = update_grid(grid)

    density = np.sum(grid, axis=1)
    busiest = np.argmax(density)

    print("Iteration:", iteration + 1)
    print("Fitness:", round(score, 4))
    print("Busiest Road:", busiest + 1)

print("\nBest Fitness:", round(best_fitness, 4))
print("Best Traffic Grid:\n", best_grid)
```
