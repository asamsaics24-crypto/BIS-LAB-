```python

import random

cities = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(cities)
population = 10
generations = 50

routes = [random.sample(range(n), n) for _ in range(population)]

for _ in range(generations):

    routes.sort(key=lambda r: sum(
        cities[r[i]][r[(i+1) % n]] for i in range(n)
    ))

    new_routes = routes[:2]

    while len(new_routes) < population:
        p1, p2 = random.sample(routes[:5], 2)

        child = p1[:2] + [x for x in p2 if x not in p1[:2]]

        if random.random() < 0.2:
            i, j = random.sample(range(n), 2)
            child[i], child[j] = child[j], child[i]

        new_routes.append(child)

    routes = new_routes

best = routes[0]
distance = sum(cities[best[i]][best[(i+1) % n]]
               for i in range(n))

print("Best Route:", best + [best[0]])
print("Minimum Distance:", distance)
```
