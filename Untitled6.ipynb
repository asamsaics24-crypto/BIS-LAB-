```python
import numpy as np
import random

# Network branches: (bus1, bus2, resistance)
branches = [
    (0, 1, 0.10), (1, 2, 0.15), (2, 3, 0.10),
    (3, 4, 0.20), (0, 4, 0.30)
]

currents = [80, 60, 50, 40, 20]
n = len(branches)

ants, iterations = 20, 100
pheromone = np.ones(n)
best_loss = float('inf')
best_solution = None

# Power loss calculation
def loss(solution):
    return sum(
        currents[i]**2 * branches[i][2]
        for i in range(n) if solution[i] == 1
    )

# ACO optimization
for iteration in range(iterations):
    for ant in range(ants):
        solution = [0] * n

        # Select four branches
        selected = random.sample(range(n), 4)

        for i in selected:
            solution[i] = 1

        value = loss(solution)

        if value < best_loss:
            best_loss = value
            best_solution = solution[:]

    # Update pheromones
    pheromone *= 0.7

    for i in range(n):
        if best_solution[i]:
            pheromone[i] += 1 / (best_loss + 1e-9)

# Display results
print("Best branch configuration:")

for i, status in enumerate(best_solution):
    print(
        f"Branch {branches[i][:2]}:",
        "Closed" if status else "Open"
    )

print("Minimum power loss:", best_loss, "W")

```
